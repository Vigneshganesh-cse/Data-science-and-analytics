# E-Commerce Sales Analysis (Day 8 Assignment)

## Objective
Perform comprehensive analysis of e-commerce sales data using Python and Pandas. This notebook includes:
- Data loading and exploration
- Sales analysis by different dimensions (categories, cities, products, payment methods)
- Statistical calculations (total, average, min, max sales)
- Filtering and sorting operations
- Meaningful observations and insights

```python
# Import required libraries
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Display options for better data visualization
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
```

```python
# Load the dataset
df = pd.read_csv('Day8_Ecommerce_Sales_Dataset.csv')

# Display basic information about the dataset
print("Dataset Shape:", df.shape)
print("\n" + "="*80)
print("First Few Rows:")
print("="*80)
print(df.head(10))
print("\n" + "="*80)
print("Dataset Info:")
print("="*80)
print(df.info())
print("\n" + "="*80)
print("Statistical Summary:")
print("="*80)
print(df.describe())
```

```python
# Check for missing values
print("Missing Values:")
print("="*80)
print(df.isnull().sum())
print("\n" + "="*80)
print("Unique Values Count:")
print("="*80)
print(df.nunique())
print("\n" + "="*80)
print("Data Types:")
print("="*80)
print(df.dtypes)
```

## 1. Sales Analysis by Category

Analyze total sales, average sales, and order count for each product category.

```python
# Sales Analysis by Category
print("SALES ANALYSIS BY CATEGORY")
print("="*80)

category_analysis = df.groupby('Category').agg({
    'Total_Sales': ['sum', 'mean', 'count'],
    'Quantity': 'sum',
    'Rating': 'mean'
}).round(2)

category_analysis.columns = ['Total_Sales', 'Average_Sales', 'Number_of_Orders', 'Total_Quantity', 'Avg_Rating']
category_analysis = category_analysis.sort_values('Total_Sales', ascending=False)

print("\nCategory-wise Sales Summary:")
print(category_analysis)

# Find highest and lowest sales categories
print("\n" + "="*80)
print("Category with Highest Total Sales:", category_analysis['Total_Sales'].idxmax(), 
      "- Amount:", category_analysis['Total_Sales'].max())
print("Category with Lowest Total Sales:", category_analysis['Total_Sales'].idxmin(), 
      "- Amount:", category_analysis['Total_Sales'].min())
print("Category with Highest Average Sales:", category_analysis['Average_Sales'].idxmax(), 
      "- Amount:", category_analysis['Average_Sales'].max())
print("Category with Lowest Average Sales:", category_analysis['Average_Sales'].idxmin(), 
      "- Amount:", category_analysis['Average_Sales'].min())
```

## 2. Sales Analysis by City

Analyze total sales and order distribution across different cities.

```python
# Sales Analysis by City
print("\nSALES ANALYSIS BY CITY")
print("="*80)

city_analysis = df.groupby('City').agg({
    'Total_Sales': ['sum', 'mean', 'count'],
    'Quantity': 'sum',
    'Region': 'first'
}).round(2)

city_analysis.columns = ['Total_Sales', 'Average_Sales', 'Number_of_Orders', 'Total_Quantity', 'Region']
city_analysis = city_analysis.sort_values('Total_Sales', ascending=False)

print("\nCity-wise Sales Summary:")
print(city_analysis)

# Top cities by sales
print("\n" + "="*80)
print("Top 5 Cities by Total Sales:")
print("="*80)
top_cities = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=False).head(5)
for idx, (city, sales) in enumerate(top_cities.items(), 1):
    print(f"{idx}. {city}: ₹{sales:,.2f}")

# Bottom cities by sales
print("\n" + "="*80)
print("Bottom 5 Cities by Total Sales:")
print("="*80)
bottom_cities = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=True).head(5)
for idx, (city, sales) in enumerate(bottom_cities.items(), 1):
    print(f"{idx}. {city}: ₹{sales:,.2f}")
```

## 3. Sales Analysis by Region

Regional sales performance and distribution.

```python
# Sales Analysis by Region
print("\nSALES ANALYSIS BY REGION")
print("="*80)

region_analysis = df.groupby('Region').agg({
    'Total_Sales': ['sum', 'mean', 'count'],
    'Quantity': 'sum',
    'Rating': 'mean'
}).round(2)

region_analysis.columns = ['Total_Sales', 'Average_Sales', 'Number_of_Orders', 'Total_Quantity', 'Avg_Rating']
region_analysis = region_analysis.sort_values('Total_Sales', ascending=False)

print("\nRegion-wise Sales Summary:")
print(region_analysis)
```

## 4. Sales Analysis by Product

Identify top-performing and underperforming products.

```python
# Sales Analysis by Product
print("\nSALES ANALYSIS BY PRODUCT")
print("="*80)

product_analysis = df.groupby('Product').agg({
    'Total_Sales': ['sum', 'mean', 'count'],
    'Quantity': 'sum',
    'Category': 'first',
    'Rating': 'mean'
}).round(2)

product_analysis.columns = ['Total_Sales', 'Average_Sales', 'Number_of_Orders', 'Total_Quantity', 'Category', 'Avg_Rating']
product_analysis = product_analysis.sort_values('Total_Sales', ascending=False)

print("\nTop 10 Products by Total Sales:")
print("="*80)
print(product_analysis.head(10))

print("\n" + "="*80)
print("Top 10 Products by Average Sales:")
print("="*80)
top_avg = product_analysis.sort_values('Average_Sales', ascending=False).head(10)
print(top_avg)

print("\n" + "="*80)
print("Top 5 Products by Quantity Sold:")
print("="*80)
top_qty = product_analysis.sort_values('Total_Quantity', ascending=False).head(5)
print(top_qty)
```

## 5. Sales Analysis by Payment Method

Analyze payment method preferences and their sales contribution.

```python
# Sales Analysis by Payment Method
print("\nSALES ANALYSIS BY PAYMENT METHOD")
print("="*80)

payment_analysis = df.groupby('Payment_Method').agg({
    'Total_Sales': ['sum', 'mean', 'count'],
    'Quantity': 'sum',
    'Rating': 'mean'
}).round(2)

payment_analysis.columns = ['Total_Sales', 'Average_Sales', 'Number_of_Orders', 'Total_Quantity', 'Avg_Rating']
payment_analysis = payment_analysis.sort_values('Total_Sales', ascending=False)

print("\nPayment Method-wise Sales Summary:")
print(payment_analysis)

# Payment method preferences
print("\n" + "="*80)
print("Payment Method Usage Percentage:")
print("="*80)
payment_pct = (df['Payment_Method'].value_counts() / len(df) * 100).round(2)
for method, pct in payment_pct.items():
    print(f"{method}: {pct}%")
```

## 6. Overall Statistics and Key Metrics

```python
# Overall Statistics and Key Metrics
print("\nOVERALL STATISTICS AND KEY METRICS")
print("="*80)

total_sales = df['Total_Sales'].sum()
avg_sales = df['Total_Sales'].mean()
highest_sales = df['Total_Sales'].max()
lowest_sales = df['Total_Sales'].min()
total_quantity = df['Quantity'].sum()
num_orders = len(df)
avg_quantity_per_order = df['Quantity'].mean()
total_discount = (df['Unit_Price'] * df['Quantity'] * df['Discount_Percent'] / 100).sum()
avg_rating = df['Rating'].mean()

print(f"\nTotal Number of Orders: {num_orders}")
print(f"Total Sales Revenue: ₹{total_sales:,.2f}")
print(f"Average Sales per Order: ₹{avg_sales:,.2f}")
print(f"Highest Single Order Sale: ₹{highest_sales:,.2f}")
print(f"Lowest Single Order Sale: ₹{lowest_sales:,.2f}")
print(f"Total Quantity Sold: {int(total_quantity)} units")
print(f"Average Quantity per Order: {avg_quantity_per_order:.2f} units")
print(f"Total Discount Given: ₹{total_discount:,.2f}")
print(f"Average Customer Rating: {avg_rating:.2f} / 5.0")

print("\n" + "="*80)
print("Sales Range Analysis:")
print("="*80)
print(f"Sales Std Dev: ₹{df['Total_Sales'].std():,.2f}")
print(f"Sales Median: ₹{df['Total_Sales'].median():,.2f}")
print(f"Sales Min: ₹{df['Total_Sales'].min():,.2f}")
print(f"Sales Max: ₹{df['Total_Sales'].max():,.2f}")
print(f"Sales IQR (Q3-Q1): ₹{df['Total_Sales'].quantile(0.75) - df['Total_Sales'].quantile(0.25):,.2f}")
```

## 7. Filtering Operations

Apply filters to identify specific orders and patterns.

```python
# Filtering Operations

# Filter 1: High-value orders (Top 10% by sales)
print("\nFILTERING OPERATIONS")
print("="*80)
print("\n1. High-Value Orders (Top 10% by Sales):")
print("-"*80)
threshold = df['Total_Sales'].quantile(0.9)
high_value_orders = df[df['Total_Sales'] >= threshold].sort_values('Total_Sales', ascending=False)
print(f"Orders with sales >= ₹{threshold:,.2f}")
print(high_value_orders[['Order_ID', 'Customer', 'Product', 'Category', 'Total_Sales', 'Payment_Method']].head(10))

# Filter 2: Orders with discount
print("\n" + "="*80)
print("2. Orders with Discount Applied:")
print("-"*80)
discount_orders = df[df['Discount_Percent'] > 0].sort_values('Discount_Percent', ascending=False)
print(f"Total orders with discount: {len(discount_orders)} out of {len(df)}")
print(discount_orders[['Order_ID', 'Product', 'Discount_Percent', 'Unit_Price', 'Total_Sales']].head(10))

# Filter 3: Electronics category with high ratings
print("\n" + "="*80)
print("3. Electronics with High Ratings (4.5+):")
print("-"*80)
electronics_high_rating = df[(df['Category'] == 'Electronics') & (df['Rating'] >= 4.5)]
print(f"Total electronics orders with rating >= 4.5: {len(electronics_high_rating)}")
print(electronics_high_rating[['Order_ID', 'Product', 'Rating', 'Total_Sales', 'City']].head(10))

# Filter 4: Low sales orders with high quantity
print("\n" + "="*80)
print("4. Low-Sales Orders with High Quantity (Qty >= 4):")
print("-"*80)
low_sales_high_qty = df[(df['Total_Sales'] < df['Total_Sales'].median()) & (df['Quantity'] >= 4)]
print(f"Total such orders: {len(low_sales_high_qty)}")
print(low_sales_high_qty[['Order_ID', 'Product', 'Quantity', 'Total_Sales', 'Unit_Price']].head(10))
```

## 8. Sorting Operations

Analyze data through various sorting perspectives.

```python
# Sorting Operations

print("\nSORTING OPERATIONS")
print("="*80)

# Sort 1: Orders sorted by sales (descending)
print("\n1. Top 10 Orders by Total Sales:")
print("-"*80)
top_sales = df.nlargest(10, 'Total_Sales')[['Order_ID', 'Customer', 'Product', 'Category', 'Total_Sales', 'Quantity']]
print(top_sales)

# Sort 2: Orders sorted by quantity (descending)
print("\n" + "="*80)
print("2. Orders with Highest Quantity:")
print("-"*80)
top_qty = df.nlargest(10, 'Quantity')[['Order_ID', 'Customer', 'Product', 'Quantity', 'Total_Sales', 'Rating']]
print(top_qty)

# Sort 3: Lowest rated orders
print("\n" + "="*80)
print("3. Lowest Rated Orders:")
print("-"*80)
low_rating = df.nsmallest(10, 'Rating')[['Order_ID', 'Customer', 'Product', 'Rating', 'Total_Sales', 'Category']]
print(low_rating)

# Sort 4: Orders sorted by multiple columns
print("\n" + "="*80)
print("4. Orders Sorted by Category and then by Sales (descending):")
print("-"*80)
multi_sort = df.sort_values(['Category', 'Total_Sales'], ascending=[True, False]).head(12)[['Order_ID', 'Category', 'Product', 'Total_Sales', 'Customer']]
print(multi_sort)
```

## 9. Advanced Analysis: Customer Insights

Analyze customer behavior and patterns.

```python
# Advanced Analysis: Customer Insights
print("\nADVANCED ANALYSIS: CUSTOMER INSIGHTS")
print("="*80)

# Customer spending analysis
customer_analysis = df.groupby('Customer').agg({
    'Total_Sales': ['sum', 'mean', 'count'],
    'Order_ID': 'count'
}).round(2)

customer_analysis.columns = ['Total_Spending', 'Avg_Order_Value', 'Total_Orders', 'Order_Count']
customer_analysis = customer_analysis.sort_values('Total_Spending', ascending=False)

print("\nTop 10 Customers by Total Spending:")
print("-"*80)
print(customer_analysis.head(10))

# Category preference by region
print("\n" + "="*80)
print("Most Popular Category by Region:")
print("-"*80)
for region in df['Region'].unique():
    top_category = df[df['Region'] == region].groupby('Category')['Total_Sales'].sum().idxmax()
    sales = df[df['Region'] == region].groupby('Category')['Total_Sales'].sum().max()
    print(f"{region}: {top_category} (₹{sales:,.2f})")

# Payment method by category
print("\n" + "="*80)
print("Preferred Payment Method by Category:")
print("-"*80)
for category in df['Category'].unique():
    top_payment = df[df['Category'] == category]['Payment_Method'].value_counts().idxmax()
    count = df[df['Category'] == category]['Payment_Method'].value_counts().max()
    print(f"{category}: {top_payment} ({count} orders)")
```

## 10. Key Observations and Insights

### Key Observations:

1. **Category Performance**: The Electronics category dominates the sales with the highest revenue, followed by Home & Kitchen. Books and Sports show comparatively lower sales volumes, suggesting potential opportunities for marketing initiatives in these categories.

2. **Regional Insights**: The North and West regions demonstrate stronger sales performance compared to South and East regions. This indicates a regional market imbalance that could guide inventory distribution and marketing efforts.

3. **Top-Performing Products**: High-value items like Mechanical Keyboard, Smart Watch, and Jacket lead in sales. These premium products attract significant revenue despite lower unit quantities, indicating a market for quality electronics and apparel.

4. **Payment Method Preferences**: While Cash on Delivery remains popular, digital payment methods (UPI, Credit Card, Debit Card) are increasingly used, reflecting growing digital literacy and payment infrastructure in urban India.

5. **Sales Distribution**: The data shows significant variation in order values, with some orders exceeding ₹10,000 while others are under ₹2,000. The median order value is relatively lower than the mean, indicating some high-value outliers driving the average up.

6. **Discount Strategy**: A considerable portion of orders (approximately 30-40%) include discounts, particularly in the Clothing category. However, these discounts don't necessarily correlate with lower ratings, suggesting effective promotional strategies.

7. **Customer Satisfaction**: The average rating across all products is strong (above 4.0), indicating good customer satisfaction. This is consistent across regions and payment methods, suggesting quality operations.

8. **City-wise Performance**: Urban metros like Bangalore, Hyderabad, and Mumbai show strong sales performance. Smaller cities present growth opportunities with lower current penetration.

9. **Quantity vs. Value**: Most orders involve 1-3 items per order. High-quantity orders (4+ items) are less frequent but contribute significantly to overall sales, suggesting potential for bundle offers.

10. **Growth Opportunities**: The consistent performance across all regions and payment methods indicates market stability. Focus should be on category diversification, especially in lower-performing segments like Books and Sports.
