# Day 13: Restaurant Branch Performance EDA

This notebook explores the provided restaurant branch performance dataset using Pandas and visual analysis. It examines descriptive statistics, distributions, branch/region/store-type comparisons, and correlations among operational and financial measures. The final section reports data-driven observations generated from the dataset.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

DATA_PATH = Path('Day13_Restaurant_Branch_Performance_Dataset.csv')
if not DATA_PATH.exists():
    raise FileNotFoundError(f'Dataset not found: {DATA_PATH.resolve()}')

df = pd.read_csv(DATA_PATH, parse_dates=['Date'])
df['Profit_Margin_Percent'] = df['Profit'].div(df['Revenue']).mul(100)

print('Dataset shape:', df.shape)
print('\nPreview:')
display(df.head())
print('\nData types:')
display(df.dtypes.to_frame('dtype'))
print('\nMissing values:')
display(df.isnull().sum().to_frame('missing_values'))
print('\nDuplicate rows:', int(df.duplicated().sum()))
print('\nCategorical value counts:')
for column in ['Branch', 'Region', 'Store_Type', 'Weather', 'Promotion']:
    print(f'\n{column}:')
    display(df[column].value_counts().to_frame('count'))

numeric_columns = [
    'Staff_Count', 'Marketing_Spend', 'Customers', 'Orders', 'Average_Bill',
    'Revenue', 'Food_Cost', 'Operating_Cost', 'Profit', 'Avg_Delivery_Min',
    'Customer_Rating', 'Temperature_C', 'Humidity_Percent', 'Profit_Margin_Percent'
]
print('\nDescriptive statistics:')
display(df[numeric_columns].describe().T.round(2))

print('\nSelected descriptive metrics:')
display(df[['Customers', 'Revenue', 'Profit', 'Marketing_Spend', 'Staff_Count', 'Avg_Delivery_Min', 'Customer_Rating']].agg(['mean', 'median', 'min', 'max']).round(2))

branch_summary = df.groupby('Branch').agg(
    Records=('Record_ID', 'count'),
    Avg_Customers=('Customers', 'mean'),
    Avg_Revenue=('Revenue', 'mean'),
    Avg_Profit=('Profit', 'mean'),
    Avg_Marketing_Spend=('Marketing_Spend', 'mean'),
    Avg_Staff_Count=('Staff_Count', 'mean'),
    Avg_Delivery_Min=('Avg_Delivery_Min', 'mean'),
    Avg_Rating=('Customer_Rating', 'mean'),
    Avg_Profit_Margin=('Profit_Margin_Percent', 'mean')
).sort_values('Avg_Profit', ascending=False)

region_summary = df.groupby('Region').agg(
    Records=('Record_ID', 'count'),
    Avg_Customers=('Customers', 'mean'),
    Avg_Revenue=('Revenue', 'mean'),
    Avg_Profit=('Profit', 'mean'),
    Avg_Delivery_Min=('Avg_Delivery_Min', 'mean'),
    Avg_Rating=('Customer_Rating', 'mean')
).sort_values('Avg_Profit', ascending=False)

store_summary = df.groupby('Store_Type').agg(
    Records=('Record_ID', 'count'),
    Avg_Customers=('Customers', 'mean'),
    Avg_Revenue=('Revenue', 'mean'),
    Avg_Profit=('Profit', 'mean'),
    Avg_Marketing_Spend=('Marketing_Spend', 'mean'),
    Avg_Delivery_Min=('Avg_Delivery_Min', 'mean'),
    Avg_Rating=('Customer_Rating', 'mean')
).sort_values('Avg_Profit', ascending=False)

print('\nPerformance by branch:')
display(branch_summary.round(2))
print('\nPerformance by region:')
display(region_summary.round(2))
print('\nPerformance by store type:')
display(store_summary.round(2))

sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(2, 2, figsize=(14, 9))
for axis, column, title in zip(
    axes.flat,
    ['Customers', 'Revenue', 'Profit', 'Customer_Rating'],
    ['Customer distribution', 'Revenue distribution', 'Profit distribution', 'Rating distribution']
):
    sns.histplot(df[column], kde=True, ax=axis, color='#176b87')
    axis.set_title(title)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
sns.barplot(data=branch_summary.reset_index(), x='Branch', y='Avg_Profit', ax=axes[0], color='#e07a5f')
axes[0].tick_params(axis='x', rotation=45)
axes[0].set_title('Average profit by branch')
axes[0].set_ylabel('Average profit')
sns.boxplot(data=df, x='Store_Type', y='Profit', ax=axes[1], color='#81b29a')
axes[1].set_title('Profit distribution by store type')
plt.tight_layout()
plt.show()

correlation_columns = [
    'Customers', 'Orders', 'Average_Bill', 'Revenue', 'Food_Cost',
    'Operating_Cost', 'Profit', 'Marketing_Spend', 'Staff_Count',
    'Avg_Delivery_Min', 'Customer_Rating', 'Profit_Margin_Percent'
]
correlation_matrix = df[correlation_columns].corr()
print('\nCorrelation matrix:')
display(correlation_matrix.round(2))

plt.figure(figsize=(12, 9))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='RdBu_r', center=0, square=True)
plt.title('Correlation matrix of numerical performance measures')
plt.tight_layout()
plt.show()
```

## Data-Driven Observations

The following observations are calculated from the grouped summaries and correlation matrix above.

```python
customer_revenue_corr = df['Customers'].corr(df['Revenue'])
marketing_profit_corr = df['Marketing_Spend'].corr(df['Profit'])
delivery_rating_corr = df['Avg_Delivery_Min'].corr(df['Customer_Rating'])
orders_revenue_corr = df['Orders'].corr(df['Revenue'])

print('1. Branch performance:')
top_branch = branch_summary.index[0]
print(f'   {top_branch} has the highest average profit at {branch_summary.loc[top_branch, "Avg_Profit"]:,.2f}, with an average margin of {branch_summary.loc[top_branch, "Avg_Profit_Margin"]:.2f}%.')

print('2. Customer volume:')
top_customers_branch = branch_summary['Avg_Customers'].idxmax()
print(f'   {top_customers_branch} serves the most customers on average ({branch_summary.loc[top_customers_branch, "Avg_Customers"]:.2f} per record).')

print('3. Regional comparison:')
top_region = region_summary.index[0]
print(f'   {top_region} leads the regions by average profit ({region_summary.loc[top_region, "Avg_Profit"]:,.2f}) and generates average revenue of {region_summary.loc[top_region, "Avg_Revenue"]:,.2f}.')

print('4. Store-type comparison:')
top_store = store_summary.index[0]
print(f'   {top_store} stores have the highest average profit ({store_summary.loc[top_store, "Avg_Profit"]:,.2f}) and average revenue ({store_summary.loc[top_store, "Avg_Revenue"]:,.2f}).')

print('5. Customer demand and revenue:')
print(f'   Customers and revenue have a correlation of {customer_revenue_corr:.2f}; orders and revenue have a correlation of {orders_revenue_corr:.2f}.')

print('6. Marketing and profitability:')
print(f'   Marketing spend and profit have a correlation of {marketing_profit_corr:.2f}, indicating the strength and direction of their linear relationship in this dataset.')

print('7. Delivery and satisfaction:')
print(f'   Average delivery time and customer rating have a correlation of {delivery_rating_corr:.2f}; this describes whether longer delivery times tend to coincide with higher or lower ratings.')

print('8. Operational scale:')
staff_customers_corr = df['Staff_Count'].corr(df['Customers'])
print(f'   Staff count and customers have a correlation of {staff_customers_corr:.2f}, showing how staffing levels relate to customer volume.')
```
