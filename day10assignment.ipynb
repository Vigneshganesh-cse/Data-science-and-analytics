```python
import pandas as pd

# Load the dataset
flight_df = pd.read_csv('Day10_Flight_Operations_Dataset.csv')

# Basic data inspection
print('Dataset shape:', flight_df.shape)
print('\nColumns:', list(flight_df.columns))
print('\nData types:\n', flight_df.dtypes)
print('\nSample rows:\n', flight_df.head(5).to_string(index=False))
print('\nNumeric summary:\n', flight_df.describe().round(2).to_string())

# Select relevant columns
selected_cols = [
    'Flight_ID', 'Airline', 'Origin', 'Destination', 'Travel_Class',
    'Passengers', 'Seat_Capacity', 'Average_Ticket_Price',
    'Delay_Minutes', 'Flight_Status', 'Weather', 'Passenger_Satisfaction'
]
print('\nSelected columns preview:\n', flight_df[selected_cols].head().to_string(index=False))

# Filtering records
high_delay = flight_df[flight_df['Delay_Minutes'] > 30][[
    'Flight_ID', 'Airline', 'Origin', 'Destination', 'Delay_Minutes', 'Weather', 'Flight_Status'
]]
print('\nFlights delayed more than 30 minutes:\n', high_delay.head(10).to_string(index=False))

# Sorting records
print('\nTop 10 longest delays:\n', flight_df.sort_values('Delay_Minutes', ascending=False)[[
    'Flight_ID', 'Airline', 'Origin', 'Destination', 'Delay_Minutes'
]].head(10).to_string(index=False))

print('\nTop 10 highest ticket prices:\n', flight_df.sort_values('Average_Ticket_Price', ascending=False)[[
    'Flight_ID', 'Airline', 'Travel_Class', 'Average_Ticket_Price'
]].head(10).to_string(index=False))

# Grouping and aggregation
airline_summary = flight_df.groupby('Airline').agg(
    Flights=('Flight_ID', 'count'),
    Avg_Fare=('Average_Ticket_Price', 'mean'),
    Avg_Delay=('Delay_Minutes', 'mean'),
    Avg_Satisfaction=('Passenger_Satisfaction', 'mean')
).sort_values('Avg_Fare', ascending=False).round(2)
print('\nAirline summary:\n', airline_summary.to_string())

weather_summary = flight_df.groupby('Weather').agg(
    Flights=('Flight_ID', 'count'),
    Avg_Delay=('Delay_Minutes', 'mean'),
    On_Time_Rate=('Flight_Status', lambda x: (x == 'On Time').mean() * 100),
    Avg_Satisfaction=('Passenger_Satisfaction', 'mean')
).sort_values('Avg_Delay', ascending=False).round(2)
print('\nWeather summary:\n', weather_summary.to_string())

class_summary = flight_df.groupby('Travel_Class').agg(
    Flights=('Flight_ID', 'count'),
    Avg_Fare=('Average_Ticket_Price', 'mean'),
    Avg_Delay=('Delay_Minutes', 'mean'),
    Avg_Satisfaction=('Passenger_Satisfaction', 'mean')
).sort_values('Avg_Satisfaction', ascending=False).round(2)
print('\nTravel class summary:\n', class_summary.to_string())

# DateTime operations
flight_df['Flight_Date'] = pd.to_datetime(flight_df['Flight_Date'])
flight_df['Flight_Month'] = flight_df['Flight_Date'].dt.month_name()
flight_df['Flight_DayOfWeek'] = flight_df['Flight_Date'].dt.day_name()
print('\nDate-derived columns preview:\n', flight_df[['Flight_ID', 'Flight_Date', 'Flight_Month', 'Flight_DayOfWeek']].head().to_string(index=False))

# Data transformation using apply()
flight_df['Delay_Category'] = flight_df['Delay_Minutes'].apply(
    lambda x: 'Severe' if x >= 60 else 'Moderate' if x >= 20 else 'Low' if x > 0 else 'None'
)

flight_df['Fare_Category'] = flight_df['Average_Ticket_Price'].apply(
    lambda x: 'High' if x >= 10000 else 'Medium' if x >= 5000 else 'Low'
)

print('\nDerived columns preview:\n', flight_df[['Flight_ID', 'Delay_Minutes', 'Delay_Category', 'Average_Ticket_Price', 'Fare_Category']].head().to_string(index=False))

# Business questions
print('\nQ1 - Highest average fare by airline:\n', flight_df.groupby('Airline')['Average_Ticket_Price'].mean().sort_values(ascending=False).head().to_string())
print('\nQ2 - Lowest average delay by airline:\n', flight_df.groupby('Airline')['Delay_Minutes'].mean().sort_values().head().to_string())
print('\nQ3 - Weather with the highest average delay:\n', flight_df.groupby('Weather')['Delay_Minutes'].mean().sort_values(ascending=False).to_string())
print('\nQ4 - Booking channel with best satisfaction:\n', flight_df.groupby('Booking_Channel')['Passenger_Satisfaction'].mean().sort_values(ascending=False).to_string())
print('\nQ5 - Flight status distribution:\n', flight_df['Flight_Status'].value_counts().to_string())

# Final cleaned analytics dataset
final_columns = [
    'Flight_ID', 'Flight_Date', 'Flight_Month', 'Flight_DayOfWeek', 'Airline', 'Origin', 'Destination',
    'Aircraft', 'Travel_Class', 'Passengers', 'Seat_Capacity', 'Average_Ticket_Price', 'Delay_Minutes',
    'Delay_Category', 'Flight_Status', 'Weather', 'Booking_Channel', 'Avg_Baggage_Kg', 'Meal_Preference',
    'Fare_Category', 'Passenger_Satisfaction'
]
cleaned_flights = flight_df[final_columns].sort_values('Flight_Date').reset_index(drop=True)
print('\nFinal dataset preview:\n', cleaned_flights.head(10).to_string(index=False))
print('\nCleaned dataset rows:', len(cleaned_flights))

# Save the final cleaned dataset for export
cleaned_flights.to_csv('flight_operations_analysis_clean.csv', index=False)
print('\nSaved cleaned dataset to flight_operations_analysis_clean.csv')
```
