## Observations about dataset

The dataset contains information about 25 students with 7 features.

The available features are Student_ID, Name, Age, Course, Marks, Attendance_Percentage, and City.

The dataset contains both numerical and categorical data. Student_ID and Age are integer columns, while Marks and Attendance_Percentage are floating-point numerical columns. Name, Course, and City are categorical/text columns.

The dataset contains 3 missing values in total. There is one missing value in the Marks column, one missing value in the Attendance_Percentage column, and one missing value in the City column.

There are no duplicate rows in the dataset.

The head(), tail(), and sample() functions were used to inspect different records. The shape, columns, and index properties were used to understand the basic structure of the dataset. The dtypes and info() functions were used to identify the data types and non-null values.

The describe() function was used to obtain descriptive statistics such as count, mean, standard deviation, minimum, maximum, and percentile values for the numerical features.

Overall, the dataset is small and easy to analyze. Before performing further analysis or machine learning, the missing values should be handled appropriately.
