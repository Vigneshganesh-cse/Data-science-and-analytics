```python
import numpy as np


students = np.array([
    "Arun",
    "Bala",
    "Chandru",
    "Deepak",
    "Ezhil"
])


marks = np.array([
    [85, 78, 92],
    [67, 72, 80],
    [90, 88, 95],
    [55, 60, 65],
    [76, 85, 82]
])

print("===== STUDENT MARKS ANALYSIS =====\n")

print("Students:")
print(students)

print("\nMarks:")
print(marks)


total_marks = np.sum(marks, axis=1)

print("\nTotal Marks:")
print(total_marks)


average_marks = np.mean(marks, axis=1)

print("\nAverage Marks:")
print(average_marks)


highest_subject = np.max(marks, axis=0)

print("\nHighest Mark in Each Subject:")
print(highest_subject)


lowest_subject = np.min(marks, axis=0)

print("\nLowest Mark in Each Subject:")
print(lowest_subject)


print("\nOverall Highest Mark:")
print(np.max(marks))


print("\nOverall Lowest Mark:")
print(np.min(marks))

print("\nOverall Median:")
print(np.median(marks))


passed_students = students[average_marks >= 75]

print("\nStudents with Average >= 75:")
print(passed_students)


students_below_75 = students[average_marks < 75]

print("\nStudents with Average < 75:")
print(students_below_75)
```

```python
top_index = np.argmax(total_marks)

print("Top Student:", students[top_index])
print("Total Marks:", total_marks[top_index])
print("Average Marks:", average_marks[top_index])
```

```python
lowest_index = np.argmin(total_marks)

print("Lowest Student:", students[lowest_index])
print("Total Marks:", total_marks[lowest_index])
print("Average Marks:", average_marks[lowest_index])
```

```python
bonus = np.array([5, 5, 5])

updated_marks = marks + bonus

print("Original Marks:")
print(marks)

print("\nBonus:")
print(bonus)

print("\nUpdated Marks:")
print(updated_marks)
```

```python
print("========== FINAL SUMMARY ==========")

print("Number of Students:", students.size)
print("Number of Subjects:", marks.shape[1])

print("\nTotal Marks of Students:")
for i in range(len(students)):
    print(students[i], ":", total_marks[i])

print("\nAverage Marks:")
for i in range(len(students)):
    print(students[i], ":", average_marks[i])

top_index = np.argmax(total_marks)

print("\nTop Student:", students[top_index])
print("Top Student Total:", total_marks[top_index])
print("Top Student Average:", average_marks[top_index])

print("\nOverall Class Average:", np.mean(marks))

print("\nStudents with Average >= 75:")
print(students[average_marks >= 75])
```
